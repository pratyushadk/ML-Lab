```python
import random


def count_pairs_with_sum(numbers, target):
    pair_count = 0

    for i in range(len(numbers)):
        for j in range(i + 1, len(numbers)):
            if numbers[i] + numbers[j] == target:
                pair_count += 1

    return pair_count


def find_range(numbers):
    if len(numbers) < 3:
        return "Range determination not possible"

    return max(numbers) - min(numbers)


def multiply_matrices(matrix_a, matrix_b):
    size = len(matrix_a)
    result = [[0 for _ in range(size)] for _ in range(size)]

    for i in range(size):
        for j in range(size):
            for k in range(size):
                result[i][j] += matrix_a[i][k] * matrix_b[k][j]

    return result


def matrix_power(matrix, power):
    size = len(matrix)

    for row in matrix:
        if len(row) != size:
            return "Matrix must be square"

    if power <= 0:
        return "Power must be a positive integer"

    result = [[0 for _ in range(size)] for _ in range(size)]

    for i in range(size):
        result[i][i] = 1

    for _ in range(power):
        result = multiply_matrices(result, matrix)

    return result


def highest_occurring_character(text):
    character_count = {}

    for character in text.lower():
        if character.isalpha():
            character_count[character] = character_count.get(character, 0) + 1

    if not character_count:
        return None, 0

    highest_character = max(character_count, key=character_count.get)

    return highest_character, character_count[highest_character]


def generate_statistics():
    numbers = [random.randint(1, 10) for _ in range(25)]

    sorted_numbers = sorted(numbers)

    mean_value = sum(numbers) / len(numbers)

    middle = len(sorted_numbers) // 2
    median_value = sorted_numbers[middle]

    frequency = {}

    for number in numbers:
        frequency[number] = frequency.get(number, 0) + 1

    mode_value = max(frequency, key=frequency.get)

    return numbers, mean_value, median_value, mode_value
```

```python
numbers = [2, 7, 4, 1, 3, 6]
pair_count = count_pairs_with_sum(numbers, 10)

print("Question 1")
print("Number of pairs with sum 10:", pair_count)


numbers = [5, 3, 8, 1, 0, 4]
range_value = find_range(numbers)

print("\nQuestion 2")
print("Range:", range_value)


matrix = [
    [1, 2],
    [3, 4]
]

power = 2
result = matrix_power(matrix, power)

print("\nQuestion 3")
print("Matrix raised to power", power, ":")

for row in result:
    print(row)


text = "hippopotamus"
character, count = highest_occurring_character(text)

print("\nQuestion 4")
print("Highest occurring character:", character)
print("Occurrence count:", count)


numbers, mean_value, median_value, mode_value = generate_statistics()

print("\nQuestion 5")
print("Generated numbers:", numbers)
print("Mean:", mean_value)
print("Median:", median_value)
print("Mode:", mode_value)
```
